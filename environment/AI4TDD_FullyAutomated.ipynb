# AI4TDD Fully Automated


Run the below cell, after the openai API key was included, such that the setup process is completed!


```python
openai_api_key = "" # @param {type:"string"}

import os
import subprocess

os.environ["PARTICIPANT_ID"] = -1
os.environ["OPEN_AI_KEY"] = openai_api_key
```

Execute the following cell, to start the agentic process.

```python
test_specifications = [
    '"TextFormatter takes arbitrary words and horizontally centers them into a line." with line width=10 and word="word"',
    '"TextFormatter takes arbitrary words and horizontally centers them into a line." with line width=10 and word="hello"',
    '"TextFormatter takes two arbitrary words and spreads them.“ with line width=10 and words=\["foo", "bar"]'
]

for test_specification in test_specifications:
    # generates the test cases from the test case specification
    result_test_case_generation = subprocess.run(["python", "ai4tdd-exp/runner.py", "--file", "ai4tdd-exp/experiment.py" "--interaction-mode", "CREATE_TEST_CODE", "--prompt", test_specification], stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)

    print(result_test_case_generation.stdout)

    print(result_test_case_generation.stderr)

    # generates the production from based on the generated error traces from the failing test cases (which will be generated automatically) and the existing test cases and production code (from the second iteration onwards)
    result_production_code_generation = subprocess.run(["python", "ai4tdd-exp/runner.py", "--file", "ai4tdd-exp/experiment.py", "--interaction-mode", "CREATE_PRODUCTION_CODE"], stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)

    print(result_production_code_generation.stdout)

    print(result_production_code_generation.stderr)
```
